# Load Tracking Data

```python
tracking_frames = []
with bz2.open('../FIFA World Cup 2022/Tracking Data/10510.jsonl.bz2', 'rt') as f:
    for line in f:
        if line.strip():
            tracking_frames.append(json.loads(line))

tracking_frames = sorted(tracking_frames, key=lambda x: x['videoTimeMs'])
tracking_df = pd.DataFrame(tracking_frames)
```

```python
# Normalize both dictionary columns across tracking_df
game_events_df = pd.json_normalize(tracking_df['game_event'])
possession_events_df = pd.json_normalize(tracking_df['possession_event'])

# Build a combined DataFrame with all requested fields
combined_events_df = pd.DataFrame({
    'game_event_type': game_events_df['game_event_type'],
    'possession_event_type': possession_events_df['possession_event_type'],
    'start_frame': game_events_df['start_frame'],
    'possession_start_frame': possession_events_df['start_frame'],
    'end_frame': game_events_df['end_frame'],
    'player_name': game_events_df['player_name'],
    'team_name': game_events_df['team_name'],
    'sequence_id': game_events_df['sequence'],
    'formatted_game_clock': game_events_df['formatted_game_clock']
})

# Filter to keep rows where game_event is not null
game_events_df = combined_events_df.dropna(subset=['game_event_type'])

# Display the result
game_events_df.iloc[700:800]
```
