# Loading libraries

```python
import json
import pandas as pd
```

```python
# Set no limit for rows
pd.set_option('display.max_rows', None)

# Set no limit for columns
pd.set_option('display.max_columns', None)

# Show up to 500 characters per cell
pd.set_option('display.max_colwidth', 500)

# Show the full content regardless of length
pd.set_option('display.max_colwidth', None)
```

# Load Metadata

```python
# 1. Load Metadata (JSON)
with open('../FIFA World Cup 2022/Metadata/10510.json', 'rt') as f:
    meta_data = json.load(f)

meta_df = pd.DataFrame(meta_data)
```
