# Load Roster Data

```python
# 1. Load RosterData (JSON)
with open('../FIFA World Cup 2022/Rosters/10510.json', 'rt') as f:
    roster_data = json.load(f)

roster_df = pd.DataFrame(roster_data)
```

```python
roster_df.info()
```

```python
print(roster_df['positionGroupType'][:5])
print(roster_df['positionGroupType'].nunique())
print(roster_df['positionGroupType'].isna().sum())
```

```python
print(roster_df['shirtNumber'][:5])
print(roster_df['shirtNumber'].nunique())
print(roster_df['shirtNumber'].isna().sum())
```

```python
print(roster_df['started'][:5])
print(roster_df['started'].nunique())
print(roster_df['started'].isna().sum())
```

```python
expanded_events_df = pd.json_normalize(roster_df['player'])
for item in expanded_events_df.columns.tolist():
    print(f"{item}: {expanded_events_df[item].isna().sum()}")
```
